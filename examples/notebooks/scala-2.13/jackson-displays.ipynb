```scala
import $ivy.`sh.almond::json-api-jackson:_`
import almond.api.AlmondJackson.Extensions._
import almond.interpreter.api.DisplayData
```

```scala
// disable pprint so that the next lines won't show any output
repl.pprinter() = repl.pprinter().copy(defaultHeight = 0)
```

```scala
case class Data(foo: String, ok: Boolean)
val data = Seq(
  Data("thing", true),
  Data("other", false)
)
```

```scala
// sends data, serialized to JSON by Jackson, with a JSON MIME type
publish.displayData("application/json", data)
```

```scala
// adds data serialized to JSON by Jackson to a DisplayData
publish.display(
  DisplayData()
    .add("text/plain", "Some data")
    .addJson("application/json", Map("count" -> data.length, "values" -> data))
)
```

```scala
// sends a map from MIME types to values, serialized to JSON by Jackson
publish.displayDataObject(
  Map(
    "application/json"               -> Map("first" -> data.head),
    "application/vnd.almond.test+json" -> Seq(1, 2, 3)
  )
)
```
